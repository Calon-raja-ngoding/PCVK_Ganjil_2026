**Pertanyaan : Apakah membalik citra secara horizontal menggunakan cv.flip() mengubah nilai rata-rata intensitas citra?**

```python
import cv2 as cv
import numpy as np
from google.colab.patches import cv2_imshow

image = cv.imread("/content/ktm.jpg")
cv2_imshow(image)

rata_asli = np.mean(image)
print("Rata-rata citra asli :", rata_asli)
print("\n")

image_flip = cv.flip(image, 1)
cv2_imshow(image_flip)

rata_flip = np.mean(image_flip)
print("Rata-rata setelah flip :", rata_flip)
```

Membalik citra secara horizontal tidak mengubah nilai rata-rata intensitas citra. Hal ini karena cv.flip() hanya mengubah posisi piksel dari kiri ke kanan tanpa mengubah nilai intensitas pikselnya. Oleh karena itu, nilai rata-rata intensitas sebelum dan sesudah citra dibalik tetap sama.

**1. Studi Kasus 1: Panitia sebuah kegiatan perlu membubuhkan label identitas pada ratusan foto dokumentasi. Foto berasal dari beberapa kamera dan ponsel, sehingga ukurannya bermacam-macam. Label harus terlihat seimbang pada semua foto tanpa disetel satu per satu. Fungsi yang sama dikenakan pada dua citra berbeda ukuran. Tinggi bar dan ukuran huruf ikut menyesuaikan, sehingga proporsinya tetap sama.**

```python
import cv2 as cv
from skimage import io
from google.colab.patches import cv2_imshow

img = cv.imread("/content/drive/MyDrive/PCVK 2026/ktm.jpg")
img2 = cv.imread("/content/drive/MyDrive/PCVK 2026/ktm2.png")
img3 = io.imread("https://iiif.lib.ncsu.edu/iiif/0052574/full/800,/0/default.jpg")
img4 = io.imread("https://iiif.lib.ncsu.edu/iiif/0016007/full/800,/0/default.jpg")
img5 = io.imread("https://berita.pesisirselatankab.go.id/asset/foto_berita/newsCover_2023_7_27_1690442617847-p3oo3l.jpeg")

arrayImg = [img, img2, img3, img4, img5]

for i in arrayImg:
  lebar = i.shape[1]
  for j in range (lebar):
    tinggi = int(i.shape[0] / 12)
    for k in range (tinggi):
      i[int(k + (i.shape[0] - tinggi)),j] = [0,0,0]
  font= cv.FONT_HERSHEY_TRIPLEX
  cv.putText(i, text='Dokumentasi PCVK - 2026', org=(int(i.shape[1]/30),int(i.shape[0]-tinggi + (tinggi * 0.6))), fontFace=font, fontScale=(tinggi/100), color=(255,255,255), thickness=max(1, int(tinggi /50)), lineType=cv.LINE_AA)
  cv2_imshow(i)
  print("\n")
```
