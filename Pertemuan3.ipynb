```python
import cv2 as cv
import numpy as np
from google.colab.patches import cv2_imshow
from matplotlib import pyplot as plt

print(' Mengubah citra menjadi grayschale ')
print('-----------------------------------')

original = cv.imread('/content/drive/MyDrive/PCVK 2026/paprika.png')
lightness = np.zeros(original.shape, original.dtype)
luminance = np.zeros(original.shape, original.dtype)
avg = np.zeros(original.shape, original.dtype)

for y in range(original.shape[0]):
  for x in range(original.shape[1]):
    for c in range(original.shape[2]):
     avg[y,x,c] = int((float(original[y,x,0]) + float(original[y,x,1]) + float(original[y,x,2])) / 3)

for y in range(original.shape[0]):
  for x in range(original.shape[1]):
    for c in range(original.shape[2]):
     lightness[y,x,c] = int((float(np.max(original[y,x])) + float(np.min(original[y,x]))) / 2)

for y in range(original.shape[0]):
  for x in range(original.shape[1]):
    for c in range(original.shape[2]):
     luminance[y,x,c] = int(float(0.21 * original[y, x, 0]) + float(0.72 * original[y, x, 1]) + float(0.07 * original[y, x, 2]))

def show_comparison(title, img_gray):
    img_rgb = cv.cvtColor(original, cv.COLOR_BGR2RGB)
    fig, axes = plt.subplots(1, 2, figsize=(10, 4))

    axes[0].imshow(img_rgb)

    axes[1].imshow(img_gray, cmap='gray', vmin=0, vmax=255)

    plt.suptitle(title, x=0.125, y=0.95, ha='left', fontsize=12)
    plt.show()

show_comparison("a. Averaging", avg)
show_comparison("b. Lightness", lightness)
show_comparison("c. Luminance", luminance)
```

```python
import cv2 as cv
import numpy as np
from google.colab.patches import cv2_imshow
from matplotlib import pyplot as plt

print(' Mengubah citra menjadi grayschale di warna tertentu ')
print('-----------------------------------------------------')

original = cv.imread('/content/drive/MyDrive/PCVK 2026/paprika.png')
avg = np.zeros(original.shape, original.dtype)

for y in range(original.shape[0]):
  for x in range(original.shape[1]):
    if (original[y,x,2] > original[y,x,0]) & (original[y,x,2] > original[y,x,1]) & (original[y,x,2] > 150) & (original[y,x,1] < 100):
      avg[y,x] = original[y,x]
    else:
      for c in range(original.shape[2]):
          avg[y,x,c] = int((float(original[y,x,0]) + float(original[y,x,1]) + float(original[y,x,2])) / 3)

def show_comparison( img_gray):
    img_rgb = cv.cvtColor(original, cv.COLOR_BGR2RGB)
    fig, axes = plt.subplots(1, 2, figsize=(10, 4))

    axes[0].imshow(img_rgb)

    axes[1].imshow(cv.cvtColor(img_gray, cv.COLOR_BGR2RGB))
    plt.show()

show_comparison( avg)
```

```python
import cv2 as cv
import numpy as np
from google.colab.patches import cv2_imshow
from matplotlib import pyplot as plt

print(' Gamma correction citra ')
print('------------------------')
try:
  gamma = float(input('Masukkan nilai Gamma: '))
except ValueError:
  print('Error, not a number')

original = cv.imread('/content/drive/MyDrive/PCVK 2026/houses.jpeg')
gamma_image = np.zeros(original.shape, original.dtype)

for y in range(original.shape[0]):
  for x in range(original.shape[1]):
    for c in range(original.shape[2]):
      gamma_image[y,x,c] = 255 * ((original[y,x,c] / 255) ** gamma)

def show_comparison(nilai, gamma_img):
    img_rgb = cv.cvtColor(original, cv.COLOR_BGR2RGB)
    fig, axes = plt.subplots(1, 2, figsize=(10, 5))

    axes[0].imshow(img_rgb)
    axes[0].set_title("Citra Asli")
    axes[0].axis("off")

    axes[1].imshow(gamma_img)
    axes[1].set_title(f"Gamma Correction (γ = {nilai})")
    axes[1].axis("off")

show_comparison(gamma, gamma_image)
```

```python
import cv2 as cv
import numpy as np
from google.colab.patches import cv2_imshow
from matplotlib import pyplot as plt

print(' Kuantisasi bit citra ')
print('----------------------')
try:
  level = int(input('Masukkan bit depth tujuan (1-8): '))
except ValueError:
  print('Error, not a number')

print('Bit depth awal   : 8 bit')
print('Bit depth tujuan :', level,'3bit')
print('Jumlah level     :', 2 ** level )

bit_depth = 255 / (2 ** level - 1)
original = cv.imread('/content/drive/MyDrive/PCVK 2026/paprika.png', cv.IMREAD_GRAYSCALE)
hasil = np.zeros(original.shape, original.dtype)

for y in range(original.shape[0]):
  for x in range(original.shape[1]):
    hasil[y,x] = round(original[y,x] / bit_depth) * bit_depth

def show_comparison(nilai, gamma_img):
    img_rgb = cv.cvtColor(original, cv.COLOR_BGR2RGB)
    fig, axes = plt.subplots(1, 2, figsize=(10, 5))

    axes[0].imshow(img_rgb, cmap='gray')
    axes[0].set_title("Grayscale 8-bit")
    axes[0].axis("off")

    axes[1].imshow(gamma_img, cmap='gray')
    axes[1].set_title(f"Grayscale {nilai}-bit")
    axes[1].axis("off")
    plt.show()

show_comparison(level, hasil)
```

```python
import cv2 as cv
import numpy as np
import glob
import os

# Membaca gambar asli
original = cv.imread('/content/drive/MyDrive/PCVK 2026/galaxy/galaxy.png')

# Membuat 100 gambar Gaussian Noise
for i in range(100):

    # Membuat Gaussian Noise
    noise = np.random.normal(0, 25, original.shape)

    # Menambahkan noise ke gambar
    noisy_image = original.astype(np.float32) + noise

    # Membatasi nilai pixel 0-255
    noisy_image = np.clip(noisy_image, 0, 255)

    # Mengubah kembali ke uint8
    noisy_image = noisy_image.astype(np.uint8)

    # Menyimpan gambar
    cv.imwrite(
        f'/content/drive/MyDrive/PCVK 2026/galaxy/galaxy{i+1}.jpg',
        noisy_image
    )

print("100 gambar Gaussian Noise berhasil dibuat!")
```

```python
import cv2 as cv
import numpy as np
from google.colab.patches import cv2_imshow
import glob
import cv2 as cv

paths = glob.glob('/content/drive/MyDrive/PCVK 2026/galaxy/*.jpg')

print(' Membuat citra menggunakan Average Denoising ')
print('---------------------------------------------')
try:
  jumlah = int(input('Masukkan jumlah gambar: '))
except ValueError:
  print('Error, not a number')

original = cv.imread('/content/drive/MyDrive/PCVK 2026/galaxy.png')
hasil = np.zeros(original.shape, original.dtype)
noisy_images = []

for j in range (jumlah):
    img = cv.imread(paths[j])
    noisy_images.append(img)

for y in range(original.shape[0]):
  for x in range(original.shape[1]):
    for c in range(original.shape[2]):
      total = 0
      for i in range (jumlah):
        total += noisy_images[i][y,x,c]
      hasil[y,x,c] = round(total / jumlah)

final_frame = np.hstack((original, hasil))
cv2_imshow(final_frame)

# print("Jumlah paths:", len(paths))
# print("Jumlah noisy images:", len(noisy_images))
# print("Original:", original.shape, original.dtype)
# print("Noisy pertama:", noisy_images[0].shape, noisy_images[0].dtype)
# print("Pixel original:", original[100,100])
# print("Pixel noisy:", noisy_images[0][100,100])

# print("Jumlah gambar:", len(paths))

# for path in paths[:5]:
#     print(path)

# print("Original:", original is not None)

# print("total:", total)
# print("hasil min:", hasil.min())
# print("hasil max:", hasil.max())
```
